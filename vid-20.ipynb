#### EDA using Univariate Analysis

```python
import seaborn as sns
import pandas as pd 

df = pd.read_csv("placement.csv")
sns.countplot(x=df["placement"])
```

### piechart

```python
df["placement"].value_counts().plot(kind="pie", autopct="%.1f")
```

### Numerical Data

#### Histogram

```python
import matplotlib.pyplot as plt
plt.hist(df["iq"])
```

#### b. bistplot

```python
sns.displot(df["iq"], kde=1)
```

#### BoxPlot

```python
sns.boxplot(x=df["iq"])
```

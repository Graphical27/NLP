### Normalization is a technique used to scale the values of features to a specific range, typically [0, 1]. This is done by subtracting the minimum value of the feature and dividing by the range (max - min). The formula for normalization is:

    x' = (x - min) / (max - min)

Where:
- x' = normalized value
- x = original value    
- min = minimum value of the feature
- max = maximum value of the feature

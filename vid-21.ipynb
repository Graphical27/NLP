#### Bivariate and Multivariate Analysis

```python
import seaborn as sns
import pandas as pd

cricket = pd.read_csv("placement.csv")

tips = sns.load_dataset("tips")

titanic = sns.load_dataset("titanic")

flights = sns.load_dataset("flights")

iris = sns.load_dataset("iris")
```

#### Bi ScatterPlot (Numerical - Numerical)

```python
sns.scatterplot(x=tips["total_bill"], y=tips["tip"], hue=tips['sex'], style=tips['smoker'], size=tips['size'])
```

#### BarPlot 

###

```python
sns.barplot(x=titanic['pclass'], y=titanic['age'], hue=titanic['sex'])
```

### BoxPlot (Numerical - cato)

```python
sns.boxplot(x=titanic['sex'],y=titanic['age'], hue=titanic['survived'])
```

#### DistPlot (Numerical - cato)

```python
sns.displot(titanic[titanic['survived'] == 0]['age'], kde=True)
sns.displot(titanic[titanic['survived'] == 1]['age'], kde=True, color='red')
```

### Heatmap (cat - cat)

```python
sns.heatmap(pd.crosstab(titanic['pclass'], titanic['survived']))
```

```python
(titanic.groupby('pclass')['survived'].mean()*100).plot(kind='bar')
```

#### CluserMAP (cat - cat) ## dendomap


```python
sns.clustermap(pd.crosstab(titanic['parch'], titanic['survived']))
```

### Pairplot

```python
iris.sample(4)
```

```python
sns.pairplot(iris, hue='species')
```

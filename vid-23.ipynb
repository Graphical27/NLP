### Feature Engineering

##### Feature engineering is the process of using domain knowledge to extract features from raw data. These features can be used to improve the performance of machine learning algorithms.

<img src="image.png" height="600">

#### Key Concepts:

**Feature Transformation** - Feature transformation is the process of mathematically modifying existing features to make them more suitable for machine learning models.

**Feature Construction** - The process of creating new features from existing raw data by applying transformations, combinations, or domain-specific logic to generate meaningful variables.

**Feature Selection** - The process of identifying and choosing the most relevant features from the available feature set to improve model performance, reduce overfitting, and decrease computational cost.

**Feature Extraction** - The process of automatically deriving new features from raw data (such as images, text, or signals) through techniques like PCA, autoencoders, or domain-specific transformations.

 Sampling noise is the random variation you get because a sample is only a small part of the full population. Even if the data is collected correctly, different samples can produce slightly different results just by chance.

Sampling bias is a systematic error that happens when the sample is not representative of the population. This means some groups or patterns are overrepresented or underrepresented, so the sample gives a distorted picture.

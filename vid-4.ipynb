##### Batch Learning: Batch learning is a training approach where a model learns from the entire dataset (or a large fixed chunk) at once, then updates its parameters.


#### Advantages of Batch Learning

Higher training quality on stable data because the model sees the full dataset.

Easier to debug and reproduce since training happens in controlled runs.

Can be optimized well with mini-batches on GPUs.

Good when data does not change quickly (for example historical records).

#### Disadvantages of Batch Learning

Slow to adapt to new trends because the model is not updated continuously.

Retraining can be expensive in time and compute for large datasets.

Needs enough storage/memory to manage large training data.

If data distribution changes between retraining cycles, performance can drop until the next retrain.

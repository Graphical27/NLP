#### Feature Scailing: Feature Scaling is a technique to standardize the independent features present in the data in a fixed range


##### use in KNN algo   

### Types of feature scailing  

    - Standardization
    - Normalization
#### Standardization: Standardization is a technique to transform the data into a standard normal distribution with a mean of 0 and a standard deviation of 1. It is also known as Z-score normalization.

#### Normalization: Normalization is a technique to transform the data into a range of [0, 1] or [-1, 1]. It is also known as Min-Max normalization.



### standardization



after standardization, the data will have a mean of 0 and a standard deviation of 1. The formula for standardization is:

    z = (x - μ) / σ

Where:
- z = standardized value
- x = original value
- μ = mean of the original values
- σ = standard deviation of the original values



| Algorithm(s) | Reason for Applying Feature Scaling |
|---|---|
| **K-Means** | Uses Euclidean distance, so scaling affects distance calculations. |
| **K-Nearest Neighbours (KNN)** | Distance between samples depends on the measurement scale of features. |
| **Principal Component Analysis (PCA)** | Looks for features with maximum variance, so scaling is important. |
| **Artificial Neural Network (ANN)** | Uses gradient descent, which benefits from scaled inputs. |
| **Gradient Descent** | Feature scaling speeds up convergence and makes learning-rate updates more stable. |

**Note:** Feature scaling is generally **not required for Decision Trees**.

#### Online Learning:Online learning is a machine learning approach where the model is updated continuously (or very frequently) as new data arrives, instead of retraining from scratch on the entire dataset.


#### Advantages of Online Learning

Adapts quickly to new data patterns (concept drift).

Works well for streaming/real-time data.

Lower memory requirement since it does not need the full dataset at once.

Can update continuously without full retraining.

#### Disadvantages of Online Learning

More sensitive to noisy or bad incoming data.

Can “forget” older patterns if updates are not controlled.

Harder to monitor/debug than scheduled batch retraining.

Requires careful tuning of learning rate and update strategy to avoid instability.

#### Inctace Based Learning VS Model-Based Learning 

---

#### Inctance Based Learning: Instance-Based Learning (IBL) is a machine learning approach where the model stores training examples and makes predictions by comparing a new input to the most similar stored examples.

Instance aka Lazy Learner

#### Advantages:

Simple and intuitive.

Little to no explicit “training” phase (fast to set up).

Can model complex decision boundaries.

Easy to update with new data (just add new instances).

Often performs well when similar inputs have similar outputs.

#### Disadvantages:

Prediction can be slow on large datasets (must compare with many instances).

High memory usage (stores lots of data).

Sensitive to irrelevant features and feature scaling.

Performance depends heavily on distance metric and choice of (k) (or neighborhood size).

Can be affected by noise/outliers in the training data.

#### Model Based Learning: Model-Based Learning is a machine learning approach where the system learns a general model (a mathematical function or set of rules) from training data, then uses that model to make predictions on new data


#### Advantages:
Faster prediction after training, since it uses a learned model instead of comparing all past examples.

Usually needs less memory than storing the full dataset.

Can generalize well to unseen data when the model is chosen correctly.

Easier to interpret in some models (for example, linear models or decision trees).

Better suited for large-scale applications with many real-time predictions.

####Disadvantages:

Requires a training phase, which can be time-consuming.

Model performance depends on choosing the right algorithm and hyperparameters.

Can underfit if the model is too simple or overfit if too complex.

Harder to update incrementally in some models (may require retraining).

If assumptions of the model are wrong, prediction quality drops.
